# Kitchen Hygiene AI — working prototype (Group 10)
A web app that inspects a kitchen CCTV clip: it finds each person and their wrists, tracks them, checks hygiene rules over time,
and explains every alert with a confidence score, an evidence frame, the annotated video and the reasoning.

**How to run (Google Colab, about 3 minutes):** Runtime → Run all. A GPU is optional (Runtime → Change runtime type → T4 GPU makes it faster).
**No files needed** — the sample kitchen clips (S01, S05, S12, H1) download automatically with the code.
Cell 3 prints a **public link** (`https://….gradio.live`): open it, pick a clip and press **Inspect this clip**.

Full code, training and evaluation: https://github.com/Avidha-047/kitchen-hygiene-ai

## 2. Sample clips (automatic)
The sample clips come with the code. Uploading your own is optional — the zones are marked for our demo kitchen camera, so other footage won't match.

```
import os, glob, shutil
clips = sorted(glob.glob('/content/app/samples/*.mp4'))
UPLOAD_MORE = False   # set to True only if you want to add extra clips of the same kitchen
if not clips or UPLOAD_MORE:
    from google.colab import files
    for name in files.upload():
        shutil.move(name, f'/content/app/samples/{name}')
print('Sample clips ready:', sorted(os.path.basename(p) for p in glob.glob('/content/app/samples/*.mp4')))
```
